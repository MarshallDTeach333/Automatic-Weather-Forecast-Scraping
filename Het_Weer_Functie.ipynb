```python

def het_weer_stad(stad):
    import requests
    from bs4 import BeautifulSoup
    import pandas as pd
    
    url = 'https://www.meteo.be/nl/' + str(stad)
    response = requests.get(url)
    html = response.text
    
    soup = BeautifulSoup(html, 'html.parser')
    dagen = soup.find_all('day-of-the-week')
    
    headers = ['Dag', 'Min.Temp (°C)', 'Max.Temp (°C)', 'Vocht (%)', 'Wind (Bft)']
    
    lijst_rijen = []
    
    for dag in dagen:
        naam = dag['name']
        mintemp = dag['mintemp']
        maxtemp = dag['maxtemp']
        vocht = dag['humiditypct']
        wind = dag['windspeedbft']
    
        dag_info = [naam, mintemp, maxtemp, vocht, wind]
        lijst_rijen.append(dag_info)
    
    tabel_weer = pd.DataFrame(data = lijst_rijen, columns = headers)
    tabel_weer = tabel_weer.replace(to_replace = '', value = 'NaN')

    # in plaats van naam van de dag willen we 'Vandaag' zien => Vandaag + Vannacht (duidelijke info)!
        # om de 1ste waarde te veranderen hebben we een filter nodig die we in '.replace()' kunnen steken!
    filt = tabel_weer.loc[0, 'Dag']
    tabel_weer = tabel_weer.replace(to_replace = filt, value = 'Vandaag').set_index(keys = 'Dag')
    
    return tabel_weer
```

```python
het_weer_stad('brussel')
```

```python
het_weer_stad('antwerpen')
```
